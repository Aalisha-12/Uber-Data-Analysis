```python
df['START']=df['START'].str.replace("Kar?chi","Karachi")
df['START']=df['START'].str.replace("R?walpindi","Rawalpindi")
df['START'].unique()
```

```python
df["STOP"].unique()
```

```python
df['STOP']=df['STOP'].str.replace("Kar?chi","Karachi")
df['STOP']=df['STOP'].str.replace("R?walpindi","Rawalpindi")
df['STOP'].unique()
```

```python
df['PURPOSE'].unique()
```

```python
df['PURPOSE']=df['PURPOSE'].str.strip(' ($)')
df['PURPOSE'].unique()
```

```python
df['START_DATE'].unique()
```

```python
df['START_DATE']=df['START_DATE'].str.replace('/','-')
df['START_DATE'].unique()
```

```python
df['END_DATE'].unique()
```

```python
df['END_DATE']=df['END_DATE'].str.replace('/','-')
df['END_DATE'].unique()
```

```python
df['START_DATE']=pd.to_datetime(df['START_DATE'])
df['END_DATE']=pd.to_datetime(df['END_DATE'])
```

```python
df.info()
```

Feature engineering using date columns

```python
df["START_DATE"].unique()
```

```python
df["DATE"]=df["START_DATE"].dt.date
df["TIME"]=df["START_DATE"].dt.hour
df
```

```python
# convert time into 4 categories

out=[]
for i in df["TIME"]:
  if 4<=i<=11:
    out.append("Morning")
  elif 12<=i<=16:
    out.append("Afternoon")
  elif 17<=i<=19:
    out.append("Evening")
  elif 20<=i<=24 or 0<=i<=3:
    out.append("Night")
df["DAY-NIGHT"]=out
df
```

```python
df["MONTH"]=df["START_DATE"].dt.month_name()
df.head()
```

```python
df["DAY"]=df["START_DATE"].dt.day_name()
df.head()
```
